# Stabilizer Logic

Welcome to your first deep system in quantum error correction: **stabilizers**.

> A stabilizer is a symmetry that a good state obeys. An error breaks the symmetry. Recovery restores it.

This notebook will walk you through what stabilizers are, how they detect errors, and why they're the foundation of fault-tolerant quantum codes.

## 1. Pauli Operators Refresher

We work with four basic single-qubit operators:
- `I = identity`
- `X = bit flip`
- `Y = bit and phase flip`
- `Z = phase flip`

Let’s define them in code:

```python
import numpy as np

I = np.array([[1, 0], [0, 1]])
X = np.array([[0, 1], [1, 0]])
Y = np.array([[0, -1j], [1j, 0]])
Z = np.array([[1, 0], [0, -1]])
```

## 2. What Is a Stabilizer?

A stabilizer is an operator $S$ such that:

$$ S|\psi\rangle = |\psi\rangle $$

That means $|\psi\rangle$ is a +1 eigenvector of $S$.

In QEC, we define a **code space** as the set of states stabilized by a group of such operators.

## 3. Stabilizers for the |000⟩ + |111⟩ Code

Let’s take the logical 0 state:

$$ |0_L⟩ = |000⟩ \quad \text{and} \quad |1_L⟩ = |111⟩ $$

This code is stabilized by:

- $Z_1 Z_2$
- $Z_2 Z_3$

Any valid state will return to itself when we apply these operators.

```python
def tensor(*ops):
    result = ops[0]
    for op in ops[1:]:
        result = np.kron(result, op)
    return result
```

Let’s apply the stabilizers to $|000⟩$ and see that they return the same state.

```python
ket0 = np.array([1, 0])
state = tensor(ket0, ket0, ket0)

Z1Z2 = tensor(Z, Z, I)
Z2Z3 = tensor(I, Z, Z)

print("Z1Z2 |000⟩ =", Z1Z2 @ state)
print("Z2Z3 |000⟩ =", Z2Z3 @ state)
```

## 4. Introducing an Error

Let’s apply a bit flip (X) to qubit 2 and see what happens when we measure the stabilizers.

```python
X2 = tensor(I, X, I)
error_state = X2 @ state

print("Z1Z2 (after error):", Z1Z2 @ error_state)
print("Z2Z3 (after error):", Z2Z3 @ error_state)
```

You’ll notice one of the stabilizers gives you `-1 × state`.

That’s how we detect where the error occurred.

This pattern of `+1/-1` is called the **syndrome**.

## ✅ Summary

- Stabilizers are symmetries of the code space
- Errors break these symmetries
- Measuring stabilizers tells us **where and what** the error was
