## Core Concept: Structure as Stability

Let’s define what we mean by **structure**.

- A *bit* is structure: `0` or `1`
- A *qubit* is structure that can rotate
- A *code* is structure that survives noise
- A *stabilizer* is structure that can correct itself

In this curriculum, we’ll work with systems that **detect their own errors**—and then recover.

### Thought experiment:
If you encode a single logical qubit as `|ψ⟩ = α|0⟩ + β|1⟩`, how would you design a system that *knows when it's wrong*?

That’s what stabilizers do.
